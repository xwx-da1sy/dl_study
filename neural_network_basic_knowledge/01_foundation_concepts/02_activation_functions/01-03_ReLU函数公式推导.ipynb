# 01-03 ReLU 函数公式推导

ReLU 是现代深度学习中最常用的隐藏层激活函数之一。它的公式非常简单，但背后的作用很重要：保留正信号，截断负信号。

## 1. 为什么需要 ReLU

Sigmoid 和 Tanh 都会把输入压缩到有限区间。当 $z$ 很大或很小时，它们的曲线会变平，导数接近 $0$，导致梯度传递变弱。

ReLU 的想法很直接：

- 如果输入是正数，就原样通过。
- 如果输入不是正数，就输出 $0$。

这样既引入了非线性，又让正半轴的梯度保持为 $1$。

## 2. 从阈值函数到 ReLU

早期感知机使用阶跃函数：

$$
\operatorname{step}(z)=
\begin{cases}
1, & z>0 \\
0, & z\le 0
\end{cases}
$$

阶跃函数能做分类，但它几乎处处导数为 $0$，不适合梯度下降训练。

ReLU 可以看成更适合优化的阈值函数：

$$
\operatorname{ReLU}(z)=
\begin{cases}
z, & z>0 \\
0, & z\le 0
\end{cases}
$$

它也可以写成更简洁的形式：

$$
\operatorname{ReLU}(z)=\max(0,z)
$$

## 3. 公式为什么是 $\max(0,z)$

因为 ReLU 的规则就是从 $0$ 和 $z$ 里面选更大的那个。

当 $z>0$ 时：

$$
\max(0,z)=z
$$

当 $z\le 0$ 时：

$$
\max(0,z)=0
$$

所以：

$$
\max(0,z)=
\begin{cases}
z, & z>0 \\
0, & z\le 0
\end{cases}
$$

这就是 ReLU 的完整定义。

## 4. 导数推导

ReLU 是分段函数，所以导数也分段讨论。

当 $z>0$ 时：

$$
\operatorname{ReLU}(z)=z
$$

因此：

$$
\operatorname{ReLU}'(z)=1
$$

当 $z<0$ 时：

$$
\operatorname{ReLU}(z)=0
$$

因此：

$$
\operatorname{ReLU}'(z)=0
$$

所以：

$$
\operatorname{ReLU}'(z)=
\begin{cases}
1, & z>0 \\
0, & z<0
\end{cases}
$$

在 $z=0$ 处，左导数为 $0$，右导数为 $1$，所以严格来说不可导。深度学习框架通常会人为约定一个值，例如 $0$，这不影响实际训练。

## 5. 为什么 ReLU 适合深层网络

ReLU 的正半轴导数是：

$$
\operatorname{ReLU}'(z)=1, \quad z>0
$$

这意味着只要神经元处于激活状态，梯度可以比较顺畅地向前一层传递。

相比之下，Sigmoid 和 Tanh 在输入绝对值很大时会进入饱和区：

$$
\sigma'(z)\approx 0
$$

$$
\tanh'(z)\approx 0
$$

因此 ReLU 常常让深层网络训练得更快、更稳定。

## 6. 死亡 ReLU 问题

ReLU 也有缺点。如果某个神经元长期满足：

$$
z\le 0
$$

那么它的输出就是：

$$
\operatorname{ReLU}(z)=0
$$

对应梯度也是：

$$
\operatorname{ReLU}'(z)=0
$$

如果这种情况一直发生，这个神经元参数很难再更新，相当于“不工作”了。这就是死亡 ReLU 问题。

常见缓解方式包括：使用较合适的学习率、合理初始化权重，或者使用 Leaky ReLU 等变体。

## 7. ReLU 的优缺点、产生原因与应用场景

### 7.1 优点

第一个优点是计算非常简单。ReLU 只需要比较 $0$ 和 $z$ 的大小：

$$
\operatorname{ReLU}(z)=\max(0,z)
$$

相比 Sigmoid 和 Tanh 中的指数运算，ReLU 的计算成本更低。

第二个优点是正半轴不会饱和。当 $z>0$ 时：

$$
\operatorname{ReLU}(z)=z
$$

$$
\operatorname{ReLU}'(z)=1
$$

这表示激活状态下的神经元可以比较稳定地传递梯度，深层网络训练时不容易像 Sigmoid 或 Tanh 那样在正半轴出现梯度饱和。

第三个优点是会产生稀疏激活。因为当 $z\le 0$ 时：

$$
\operatorname{ReLU}(z)=0
$$

这意味着一部分神经元在某些样本上不会被激活，网络表示会更稀疏，有时有助于降低冗余表达。

### 7.2 缺点

第一个缺点是死亡 ReLU 问题。如果某个神经元长期满足：

$$
z\le 0
$$

那么它的输出和梯度都是 $0$：

$$
\operatorname{ReLU}(z)=0, \quad \operatorname{ReLU}'(z)=0
$$

这个神经元就很难继续更新。

第二个缺点是输出不是零中心。ReLU 的输出范围是：

$$
\operatorname{ReLU}(z)\in[0,+\infty)
$$

输出没有负值，这一点和 Sigmoid 类似，可能会让下一层接收到的输入整体偏正。

第三个缺点是正半轴无上界。当 $z$ 很大时：

$$
\operatorname{ReLU}(z)=z
$$

如果权重初始化、学习率或数据尺度不合适，激活值可能变得很大，影响训练稳定性。因此 ReLU 常和合适的初始化、归一化方法一起使用。

### 7.3 这些优缺点为什么会产生

ReLU 的优点和缺点都来自它的分段线性结构：

$$
\operatorname{ReLU}(z)=
\begin{cases}
z, & z>0 \\
0, & z\le 0
\end{cases}
$$

正半轴是直线，所以导数恒为 $1$：

$$
\operatorname{ReLU}'(z)=1, \quad z>0
$$

这就是它梯度传递更顺畅、适合深层网络的原因。

负半轴是常数 $0$，所以导数恒为 $0$：

$$
\operatorname{ReLU}'(z)=0, \quad z<0
$$

这就是死亡 ReLU 问题产生的根本原因。

另外，ReLU 没有像 Sigmoid 和 Tanh 那样把输出压缩到有限区间，所以正半轴不会饱和；但也正因为没有上界，激活值可能变得很大。

### 7.4 应用场景

第一个场景是普通前馈神经网络的隐藏层：

$$
\mathbf{h}=\operatorname{ReLU}(\mathbf{W}\mathbf{x}+\mathbf{b})
$$

第二个场景是 CNN 的卷积层之后。卷积提取局部特征后，通常接 ReLU 引入非线性：

$$
\mathbf{A}=\operatorname{ReLU}(\operatorname{Conv}(\mathbf{X}))
$$

第三个场景是深层网络的默认隐藏层激活函数。实际建模时，如果没有特殊原因，MLP 和 CNN 的隐藏层通常可以先从 ReLU 开始。

如果训练中发现大量神经元输出一直为 $0$，可以考虑 Leaky ReLU：

$$
\operatorname{LeakyReLU}(z)=
\begin{cases}
z, & z>0 \\
\alpha z, & z\le 0
\end{cases}
$$

其中 $\alpha$ 是一个很小的正数，例如 $0.01$。

### 7.5 Leaky ReLU、PReLU、RReLU

ReLU 变体的共同目标，是给负半轴也留一点梯度，避免普通 ReLU 在 $z<0$ 时梯度完全为 $0$。

#### Leaky ReLU

Leaky ReLU 的公式是：

$$
\operatorname{LeakyReLU}(z)=
\begin{cases}
z, & z>0 \\
\alpha z, & z\le 0
\end{cases}
$$

其中 $\alpha$ 是人为指定的小正数，常见取值如：

$$
\alpha=0.01
$$

它的导数是：

$$
\operatorname{LeakyReLU}'(z)=
\begin{cases}
1, & z>0 \\
\alpha, & z<0
\end{cases}
$$

产生原因：普通 ReLU 的负半轴导数为 $0$，Leaky ReLU 把负半轴从水平线改成一条斜率很小的直线，让负半轴也能反向传播梯度。

优点是可以缓解死亡 ReLU；缺点是 $\alpha$ 需要手动指定，不同任务中最合适的值不一定相同。

在 PyTorch 中可以写成：

```python
nn.LeakyReLU(negative_slope=0.01)
```

#### PReLU

PReLU 的全称是 Parametric ReLU，也就是带参数的 ReLU。它的公式和 Leaky ReLU 很像：

$$
\operatorname{PReLU}(z)=
\begin{cases}
z, & z>0 \\
a z, & z\le 0
\end{cases}
$$

区别在于，$a$ 不是人为固定的超参数，而是模型通过训练学出来的参数：

$$
a \leftarrow a - \eta \frac{\partial \mathcal{L}}{\partial a}
$$

产生原因：Leaky ReLU 需要手动指定负半轴斜率，PReLU 希望让模型自己学习这个斜率。

优点是更灵活，模型可以根据数据自动决定负半轴保留多少信息；缺点是引入了额外参数，数据较少时可能增加过拟合风险。

在 PyTorch 中可以写成：

```python
nn.PReLU(num_parameters=1)
```

如果希望每个通道有自己的负半轴斜率，可以设置：

```python
nn.PReLU(num_parameters=channels)
```

#### RReLU

RReLU 的全称是 Randomized Leaky ReLU，也就是随机负斜率 ReLU。训练时，它的负半轴斜率不是固定值，而是从一个区间中随机采样：

$$
\operatorname{RReLU}(z)=
\begin{cases}
z, & z>0 \\
a z, & z\le 0
\end{cases}
$$

其中训练阶段：

$$
a \sim U(l, u)
$$

$U(l,u)$ 表示在区间 $(l,u)$ 上均匀采样。推理阶段通常使用区间均值：

$$
a=\frac{l+u}{2}
$$

产生原因：RReLU 希望通过随机负斜率引入一点正则化效果，让模型不要过度依赖某个固定斜率。

优点是可以缓解死亡 ReLU，并可能带来一定正则化；缺点是训练行为带随机性，理解和复现实验时要注意随机种子。

在 PyTorch 中可以写成：

```python
nn.RReLU(lower=0.125, upper=0.333)
```

#### 三者怎么选

| 函数 | 负半轴斜率 | 主要优点 | 注意点 |
|---|---|---|---|
| Leaky ReLU | 手动固定 $\alpha$ | 简单，缓解死亡 ReLU | 需要调 $\alpha$ |
| PReLU | 学习得到 $a$ | 更灵活，能适应数据 | 增加参数，可能过拟合 |
| RReLU | 训练时随机采样 $a$ | 有一定正则化效果 | 训练带随机性 |

实用记忆：先用 ReLU；如果出现死亡 ReLU，再试 Leaky ReLU；如果数据量足够、想让斜率自动学习，可以试 PReLU；如果想加入一点随机正则化，可以试 RReLU。

````python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-5, 5, 500)
relu = np.maximum(0, z)
relu_grad = np.where(z > 0, 1, 0)

alpha_leaky = 0.1
alpha_prelu_example = 0.25
alpha_rrelu_example = 0.18

leaky_relu = np.where(z > 0, z, alpha_leaky * z)
prelu_example = np.where(z > 0, z, alpha_prelu_example * z)
rrelu_example = np.where(z > 0, z, alpha_rrelu_example * z)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(z, relu, label='ReLU')
axes[0].plot(z, leaky_relu, linestyle='--', label='Leaky ReLU, alpha=0.10')
axes[0].plot(z, prelu_example, linestyle='-.', label='PReLU example, a=0.25')
axes[0].plot(z, rrelu_example, linestyle=':', label='RReLU example, a=0.18')
axes[0].axhline(0, color='black', linewidth=0.8)
axes[0].axvline(0, color='black', linewidth=0.8)
axes[0].set_title('ReLU Variants')
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(z, relu_grad, color='#DC2626', label="ReLU derivative")
axes[1].set_ylim(-0.1, 1.1)
axes[1].set_title('Derivative of ReLU')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

## 8. 在神经网络中的使用

普通前馈神经网络的隐藏层经常写成：

$$
\mathbf{h}=\operatorname{ReLU}(\mathbf{W}\mathbf{x}+\mathbf{b})
$$

在 PyTorch 中对应：

```python
nn.Linear(in_features, out_features)
nn.ReLU()
```

先记住一个实用规则：如果是普通隐藏层，不知道选什么激活函数时，优先尝试 ReLU。
