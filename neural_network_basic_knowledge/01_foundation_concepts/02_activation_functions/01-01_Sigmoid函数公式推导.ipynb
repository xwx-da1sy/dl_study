# 01-01 Sigmoid 函数公式推导

Sigmoid 是二分类问题中最经典的激活函数之一。它的核心作用是：把任意实数 $z$ 映射到 $(0, 1)$，从而可以解释成概率。

## 1. 为什么需要 Sigmoid

在线性模型中，我们通常先得到一个原始分数：

$$
z = \mathbf{w}^{T}\mathbf{x}+b
$$

但是 $z$ 的取值范围是整个实数轴：

$$
z \in (-\infty,+\infty)
$$

如果做二分类，我们希望输出的是类别 $1$ 的概率：

$$
0 < P(y=1\mid \mathbf{x}) < 1
$$

所以问题变成：如何把任意实数 $z$ 转成 $(0,1)$ 之间的概率？Sigmoid 就是一个自然选择。

## 2. 从几率 odds 推导 Sigmoid

设类别 $1$ 的概率为：

$$
p=P(y=1\mid \mathbf{x})
$$

类别 $0$ 的概率就是：

$$
1-p=P(y=0\mid \mathbf{x})
$$

几率 odds 定义为：

$$
\operatorname{odds}=\frac{p}{1-p}
$$

它表示“属于类别 $1$ 的可能性”和“属于类别 $0$ 的可能性”的比值。

几率一定是正数：

$$
\frac{p}{1-p} > 0
$$

为了让线性模型的输出 $z$ 能对应任意正数几率，我们令：

$$
\frac{p}{1-p}=e^{z}
$$

其中 $e^z$ 的好处是：无论 $z$ 是多少，$e^z$ 都大于 $0$。

## 3. 解出概率形式

从下面这个式子开始：

$$
\frac{p}{1-p}=e^{z}
$$

两边同时乘以 $1-p$：

$$
p=e^{z}(1-p)
$$

展开右边：

$$
p=e^{z}-e^{z}p
$$

把含有 $p$ 的项移到左边：

$$
p+e^{z}p=e^{z}
$$

提取 $p$：

$$
p(1+e^{z})=e^{z}
$$

两边除以 $1+e^z$：

$$
p=\frac{e^{z}}{1+e^{z}}
$$

分子分母同时除以 $e^z$：

$$
p=\frac{1}{1+e^{-z}}
$$

于是得到 Sigmoid 公式：

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

## 4. 从 logit 角度理解

对几率取对数，得到 logit：

$$
\operatorname{logit}(p)=\log\frac{p}{1-p}
$$

逻辑回归的核心假设就是：概率的 logit 是输入特征的线性函数。

$$
\log\frac{p}{1-p}=\mathbf{w}^{T}\mathbf{x}+b
$$

令 $z=\mathbf{w}^{T}\mathbf{x}+b$，就有：

$$
p=\sigma(z)
$$

所以 Sigmoid 不是凭空出现的，它可以看成是 logit 函数的反函数。

## 5. 导数推导

Sigmoid 的导数在反向传播中非常重要。先写出：

$$
\sigma(z)=\frac{1}{1+e^{-z}}=(1+e^{-z})^{-1}
$$

对 $z$ 求导：

$$
\sigma'(z)=-(1+e^{-z})^{-2}\cdot(-e^{-z})
$$

整理得：

$$
\sigma'(z)=\frac{e^{-z}}{(1+e^{-z})^2}
$$

又因为：

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

$$
1-\sigma(z)=1-\frac{1}{1+e^{-z}}=\frac{e^{-z}}{1+e^{-z}}
$$

两者相乘：

$$
\sigma(z)(1-\sigma(z))=\frac{1}{1+e^{-z}}\cdot\frac{e^{-z}}{1+e^{-z}}
$$

$$
\sigma(z)(1-\sigma(z))=\frac{e^{-z}}{(1+e^{-z})^2}
$$

所以：

$$
\sigma'(z)=\sigma(z)(1-\sigma(z))
$$

## 6. 形状与问题

Sigmoid 有三个关键性质：

$$
\lim_{z\to +\infty}\sigma(z)=1
$$

$$
\lim_{z\to -\infty}\sigma(z)=0
$$

$$
\sigma(0)=\frac{1}{2}
$$

当 $z$ 很大或很小时，Sigmoid 曲线会变得很平，导数接近 $0$。这会让梯度在反向传播中变弱，也就是常说的梯度饱和。

## 7. Sigmoid 的优缺点以及产生原因

### 7.1 优点

第一个优点是输出可以解释成概率。因为 Sigmoid 的输出范围是：

$$
0 < \sigma(z) < 1
$$

所以在二分类任务中，我们可以把它理解为样本属于类别 $1$ 的概率：

$$
\hat{y}=P(y=1\mid \mathbf{x})=\sigma(\mathbf{w}^{T}\mathbf{x}+b)
$$

第二个优点是函数连续、平滑、可导。相比感知机里的阶跃函数：

$$
\operatorname{step}(z)=
\begin{cases}
1, & z>0 \\
0, & z\le 0
\end{cases}
$$

Sigmoid 更适合梯度下降，因为它几乎处处可导，并且导数形式很简单：

$$
\sigma'(z)=\sigma(z)(1-\sigma(z))
$$

第三个优点是它能把线性输出变成非线性输出。线性部分是：

$$
z=\mathbf{w}^{T}\mathbf{x}+b
$$

经过 Sigmoid 后变成：

$$
a=\sigma(z)
$$

这让神经网络不再只是简单的线性变换。

### 7.2 缺点

第一个缺点是容易产生梯度饱和。Sigmoid 的导数是：

$$
\sigma'(z)=\sigma(z)(1-\sigma(z))
$$

因为 $0<\sigma(z)<1$，所以导数最大值出现在 $\sigma(z)=0.5$ 时：

$$
\max \sigma'(z)=0.5\times(1-0.5)=0.25
$$

当 $z\to +\infty$ 时：

$$
\sigma(z)\to 1, \quad \sigma'(z)\to 0
$$

当 $z\to -\infty$ 时：

$$
\sigma(z)\to 0, \quad \sigma'(z)\to 0
$$

这意味着神经元一旦进入两端饱和区，反向传播得到的梯度就会非常小，参数更新会变慢。

第二个缺点是输出不是以 $0$ 为中心。Sigmoid 的输出恒为正：

$$
\sigma(z)\in(0,1)
$$

如果某一层的输出总是正数，下一层接收到的输入也会整体偏正，这可能让梯度更新方向不够均衡，优化过程更容易来回摆动。

第三个缺点是指数计算相对更贵。Sigmoid 需要计算：

$$
e^{-z}
$$

在现代硬件上这通常不是最大问题，但和 ReLU 的 $\max(0,z)$ 相比，计算仍然更复杂。

### 7.3 这些优缺点为什么会产生

Sigmoid 的优点来自它的构造目标：我们希望把任意实数 $z$ 映射成概率，所以使用了指数和归一化形式：

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

这个形式天然保证输出在 $(0,1)$ 之间，也让它适合二分类概率建模。

但缺点也来自同一个形式。为了把无限实数轴压缩到有限区间 $(0,1)$，函数两端必须逐渐变平：

$$
(-\infty,+\infty) \rightarrow (0,1)
$$

曲线变平，就意味着导数变小；导数变小，反向传播中的梯度就会变小。这就是梯度饱和产生的根本原因。

另外，Sigmoid 的输出下界是 $0$，上界是 $1$，中点是 $0.5$，不是 $0$：

$$
\sigma(0)=0.5
$$

因此它天然不是零中心函数，这就是输出偏正的根本原因。

一句话总结：Sigmoid 为了得到概率解释，牺牲了一部分深层网络中的优化效率。

```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-8, 8, 500)
sigmoid = 1 / (1 + np.exp(-z))
sigmoid_grad = sigmoid * (1 - sigmoid)

plt.figure(figsize=(8, 4))
plt.plot(z, sigmoid, label=r'$\sigma(z)$')
plt.plot(z, sigmoid_grad, label=r"$\sigma'(z)$")
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Sigmoid and Its Derivative')
plt.xlabel('z')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 8. 在神经网络中的使用

二分类模型常写成：

$$
\hat{y}=\sigma(\mathbf{w}^{T}\mathbf{x}+b)
$$

如果 $\hat{y}$ 表示类别 $1$ 的概率，那么类别预测可以写成：

$$
\operatorname{class}(\hat{y})=
\begin{cases}
1, & \hat{y}\ge 0.5 \\
0, & \hat{y}<0.5
\end{cases}
$$

实践中，Sigmoid 常用于二分类输出层；隐藏层一般优先使用 ReLU 或它的变体。
