# 01-02 Tanh 函数公式推导

Tanh 的全称是 hyperbolic tangent，也就是双曲正切函数。它和 Sigmoid 很像，但输出范围是 $(-1,1)$，并且以 $0$ 为中心。

## 1. 为什么需要 Tanh

Sigmoid 的输出范围是 $(0,1)$，所有输出都是正数。这在隐藏层中有时不够理想，因为中间特征整体偏正，会影响后续层的优化。

Tanh 希望做到两件事：

1. 保留类似 Sigmoid 的非线性压缩能力。
2. 让输出以 $0$ 为中心。

因此 Tanh 的输出范围是：

$$
-1 < \tanh(z) < 1
$$

## 2. 从双曲函数定义推导

双曲正弦定义为：

$$
\sinh(z)=\frac{e^{z}-e^{-z}}{2}
$$

双曲余弦定义为：

$$
\cosh(z)=\frac{e^{z}+e^{-z}}{2}
$$

双曲正切定义为二者相除：

$$
\tanh(z)=\frac{\sinh(z)}{\cosh(z)}
$$

代入 $\sinh(z)$ 和 $\cosh(z)$：

$$
\tanh(z)=\frac{\frac{e^{z}-e^{-z}}{2}}{\frac{e^{z}+e^{-z}}{2}}
$$

上下同时约去 $\frac{1}{2}$：

$$
\tanh(z)=\frac{e^{z}-e^{-z}}{e^{z}+e^{-z}}
$$

## 3. 和 Sigmoid 的关系

Sigmoid 是：

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

Tanh 可以由 Sigmoid 平移和缩放得到：

$$
\tanh(z)=2\sigma(2z)-1
$$

推导如下：

$$
2\sigma(2z)-1=\frac{2}{1+e^{-2z}}-1
$$

$$
2\sigma(2z)-1=\frac{2-(1+e^{-2z})}{1+e^{-2z}}
$$

$$
2\sigma(2z)-1=\frac{1-e^{-2z}}{1+e^{-2z}}
$$

分子分母同时乘以 $e^z$：

$$
2\sigma(2z)-1=\frac{e^z-e^{-z}}{e^z+e^{-z}}
$$

所以：

$$
\tanh(z)=2\sigma(2z)-1
$$

## 4. 导数推导

从定义出发：

$$
\tanh(z)=\frac{\sinh(z)}{\cosh(z)}
$$

需要知道两个导数：

$$
\frac{d}{dz}\sinh(z)=\cosh(z)
$$

$$
\frac{d}{dz}\cosh(z)=\sinh(z)
$$

使用商法则：

$$
\frac{d}{dz}\tanh(z)=\frac{\cosh^2(z)-\sinh^2(z)}{\cosh^2(z)}
$$

双曲函数有恒等式：

$$
\cosh^2(z)-\sinh^2(z)=1
$$

所以：

$$
\frac{d}{dz}\tanh(z)=\frac{1}{\cosh^2(z)}
$$

又因为：

$$
\tanh^2(z)=\frac{\sinh^2(z)}{\cosh^2(z)}
$$

$$
1-\tanh^2(z)=\frac{\cosh^2(z)-\sinh^2(z)}{\cosh^2(z)}=\frac{1}{\cosh^2(z)}
$$

因此：

$$
\frac{d}{dz}\tanh(z)=1-\tanh^2(z)
$$

## 5. 形状与性质

Tanh 有下面几个重要性质：

$$
\lim_{z\to +\infty}\tanh(z)=1
$$

$$
\lim_{z\to -\infty}\tanh(z)=-1
$$

$$
\tanh(0)=0
$$

它是奇函数：

$$
\tanh(-z)=-\tanh(z)
$$

因为输出以 $0$ 为中心，Tanh 在隐藏层中通常比 Sigmoid 更舒服一些。但当 $|z|$ 很大时，导数 $1-\tanh^2(z)$ 仍然会接近 $0$，所以它也有梯度饱和问题。

```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-5, 5, 500)
tanh = np.tanh(z)
tanh_grad = 1 - tanh ** 2
sigmoid = 1 / (1 + np.exp(-z))

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(z, sigmoid, label='Sigmoid')
axes[0].plot(z, tanh, label='Tanh')
axes[0].axhline(0, color='black', linewidth=0.8)
axes[0].axvline(0, color='black', linewidth=0.8)
axes[0].set_title('Sigmoid vs Tanh')
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(z, tanh_grad, color='#DC2626', label=r"$1-\tanh^2(z)$")
axes[1].set_title('Derivative of Tanh')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. 在神经网络中的使用

Tanh 曾经常用于隐藏层，尤其是早期的 RNN。原因是隐藏状态常需要表达正向和负向的信号，而 Tanh 的输出范围 $(-1,1)$ 正好适合这种表示。

一个隐藏层可以写成：

$$
\mathbf{h}=\tanh(\mathbf{W}\mathbf{x}+\mathbf{b})
$$

现在训练普通深层前馈网络时，隐藏层通常优先选择 ReLU；但理解 Tanh 很重要，因为它在循环神经网络、门控结构和一些归一化输出场景里仍然会出现。
