# 01-02 Tanh 函数公式推导

Tanh 的全称是 hyperbolic tangent，也就是双曲正切函数。它和 Sigmoid 很像，但输出范围是 $(-1,1)$，并且以 $0$ 为中心。

## 1. 为什么需要 Tanh

Sigmoid 的输出范围是 $(0,1)$，所有输出都是正数。这在隐藏层中有时不够理想，因为中间特征整体偏正，会影响后续层的优化。

Tanh 希望做到两件事：

1. 保留类似 Sigmoid 的非线性压缩能力。
2. 让输出以 $0$ 为中心。

因此 Tanh 的输出范围是：

$$
-1 < \tanh(z) < 1
$$

## 2. 从双曲函数定义推导

双曲正弦定义为：

$$
\sinh(z)=\frac{e^{z}-e^{-z}}{2}
$$

双曲余弦定义为：

$$
\cosh(z)=\frac{e^{z}+e^{-z}}{2}
$$

双曲正切定义为二者相除：

$$
\tanh(z)=\frac{\sinh(z)}{\cosh(z)}
$$

代入 $\sinh(z)$ 和 $\cosh(z)$：

$$
\tanh(z)=\frac{\frac{e^{z}-e^{-z}}{2}}{\frac{e^{z}+e^{-z}}{2}}
$$

上下同时约去 $\frac{1}{2}$：

$$
\tanh(z)=\frac{e^{z}-e^{-z}}{e^{z}+e^{-z}}
$$

## 3. 和 Sigmoid 的关系

Sigmoid 是：

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

Tanh 可以由 Sigmoid 平移和缩放得到：

$$
\tanh(z)=2\sigma(2z)-1
$$

推导如下：

$$
2\sigma(2z)-1=\frac{2}{1+e^{-2z}}-1
$$

$$
2\sigma(2z)-1=\frac{2-(1+e^{-2z})}{1+e^{-2z}}
$$

$$
2\sigma(2z)-1=\frac{1-e^{-2z}}{1+e^{-2z}}
$$

分子分母同时乘以 $e^z$：

$$
2\sigma(2z)-1=\frac{e^z-e^{-z}}{e^z+e^{-z}}
$$

所以：

$$
\tanh(z)=2\sigma(2z)-1
$$

## 4. 导数推导

从定义出发：

$$
\tanh(z)=\frac{\sinh(z)}{\cosh(z)}
$$

需要知道两个导数：

$$
\frac{d}{dz}\sinh(z)=\cosh(z)
$$

$$
\frac{d}{dz}\cosh(z)=\sinh(z)
$$

使用商法则：

$$
\frac{d}{dz}\tanh(z)=\frac{\cosh^2(z)-\sinh^2(z)}{\cosh^2(z)}
$$

双曲函数有恒等式：

$$
\cosh^2(z)-\sinh^2(z)=1
$$

所以：

$$
\frac{d}{dz}\tanh(z)=\frac{1}{\cosh^2(z)}
$$

又因为：

$$
\tanh^2(z)=\frac{\sinh^2(z)}{\cosh^2(z)}
$$

$$
1-\tanh^2(z)=\frac{\cosh^2(z)-\sinh^2(z)}{\cosh^2(z)}=\frac{1}{\cosh^2(z)}
$$

因此：

$$
\frac{d}{dz}\tanh(z)=1-\tanh^2(z)
$$

## 5. 形状与性质

Tanh 有下面几个重要性质：

$$
\lim_{z\to +\infty}\tanh(z)=1
$$

$$
\lim_{z\to -\infty}\tanh(z)=-1
$$

$$
\tanh(0)=0
$$

它是奇函数：

$$
\tanh(-z)=-\tanh(z)
$$

因为输出以 $0$ 为中心，Tanh 在隐藏层中通常比 Sigmoid 更舒服一些。但当 $|z|$ 很大时，导数 $1-\tanh^2(z)$ 仍然会接近 $0$，所以它也有梯度饱和问题。

## 6. Tanh 的优缺点、产生原因与应用场景

### 6.1 优点

第一个优点是输出以 $0$ 为中心。Tanh 的输出范围是：

$$
-1 < \tanh(z) < 1
$$

并且：

$$
\tanh(0)=0
$$

这意味着 Tanh 的输出既可以是正数，也可以是负数。相比 Sigmoid 的输出范围 $(0,1)$，Tanh 更适合在隐藏层中表达“正向影响”和“负向影响”。

第二个优点是非线性更强，且在 $z=0$ 附近梯度较大。Tanh 的导数是：

$$
\frac{d}{dz}\tanh(z)=1-\tanh^2(z)
$$

当 $z=0$ 时：

$$
\tanh(0)=0
$$

$$
\tanh'(0)=1-0^2=1
$$

也就是说，在输入靠近 $0$ 的区域，Tanh 的梯度比较充足，参数更新会比较明显。

第三个优点是它和 Sigmoid 有紧密关系，可以看成 Sigmoid 的平移和缩放版本：

$$
\tanh(z)=2\sigma(2z)-1
$$

这个关系说明 Tanh 保留了 Sigmoid 的 S 形非线性，同时把输出中心从 $0.5$ 移到了 $0$。

### 6.2 缺点

第一个缺点是仍然会产生梯度饱和。当 $z\to +\infty$ 时：

$$
\tanh(z)\to 1
$$

$$
\tanh'(z)=1-\tanh^2(z)\to 0
$$

当 $z\to -\infty$ 时：

$$
\tanh(z)\to -1
$$

$$
\tanh'(z)=1-\tanh^2(z)\to 0
$$

所以 Tanh 虽然比 Sigmoid 更零中心，但两端依然会变平，深层网络中仍可能出现梯度变小的问题。

第二个缺点是计算比 ReLU 更复杂。Tanh 需要指数运算：

$$
\tanh(z)=\frac{e^{z}-e^{-z}}{e^{z}+e^{-z}}
$$

而 ReLU 只需要：

$$
\operatorname{ReLU}(z)=\max(0,z)
$$

所以在普通深层前馈网络中，ReLU 通常训练更快。

第三个缺点是输出被限制在 $(-1,1)$。这在某些场景是优点，但如果模型需要保留较大的正响应，Tanh 会把它压缩掉：

$$
\lim_{z\to +\infty}\tanh(z)=1
$$

这会让强信号之间的差别变得不明显。

### 6.3 这些优缺点为什么会产生

Tanh 的优点来自它的对称结构：

$$
\tanh(z)=\frac{e^{z}-e^{-z}}{e^{z}+e^{-z}}
$$

当 $z$ 变成 $-z$ 时，分子符号会反过来，分母保持为正，因此：

$$
\tanh(-z)=-\tanh(z)
$$

这就是 Tanh 输出零中心、正负对称的根本原因。

Tanh 的缺点则来自压缩映射本身。它要把整个实数轴压缩到有限区间：

$$
(-\infty,+\infty) \rightarrow (-1,1)
$$

既然输出不能超过 $1$，也不能小于 $-1$，曲线两端就必须逐渐变平；曲线变平，导数自然趋近于 $0$。这就是 Tanh 梯度饱和的根本原因。

### 6.4 应用场景

第一个场景是 RNN 的隐藏状态。在循环神经网络中，隐藏状态常写成：

$$
\mathbf{h}_t=\tanh(\mathbf{W}_{xh}\mathbf{x}_t+\mathbf{W}_{hh}\mathbf{h}_{t-1}+\mathbf{b}_h)
$$

这里使用 Tanh，是因为隐藏状态往往需要表达正负两种方向的特征，而且输出范围 $(-1,1)$ 可以防止隐藏状态数值无限增大。

第二个场景是 LSTM / GRU 这类门控循环网络中的候选状态。例如 LSTM 的候选记忆可以写成：

$$
\tilde{\mathbf{c}}_t=\tanh(\mathbf{W}_c[\mathbf{h}_{t-1},\mathbf{x}_t]+\mathbf{b}_c)
$$

这里 Tanh 用来生成范围受控、可正可负的候选信息。

第三个场景是输出需要限制在 $(-1,1)$ 的任务。例如某些控制任务、生成模型或归一化回归任务中，如果目标值本来就在 $[-1,1]$ 附近，就可以考虑在输出层使用 Tanh：

$$
\hat{y}=\tanh(z)
$$

不过，在普通 MLP 或 CNN 的隐藏层中，现在更常用 ReLU 或 ReLU 的变体；Tanh 更适合需要零中心、正负对称、范围受控的场景。

```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-5, 5, 500)
tanh = np.tanh(z)
tanh_grad = 1 - tanh ** 2
sigmoid = 1 / (1 + np.exp(-z))

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(z, sigmoid, label='Sigmoid')
axes[0].plot(z, tanh, label='Tanh')
axes[0].axhline(0, color='black', linewidth=0.8)
axes[0].axvline(0, color='black', linewidth=0.8)
axes[0].set_title('Sigmoid vs Tanh')
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(z, tanh_grad, color='#DC2626', label=r"$1-\tanh^2(z)$")
axes[1].set_title('Derivative of Tanh')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 7. 在神经网络中的使用

Tanh 曾经常用于隐藏层，尤其是早期的 RNN。原因是隐藏状态常需要表达正向和负向的信号，而 Tanh 的输出范围 $(-1,1)$ 正好适合这种表示。

一个隐藏层可以写成：

$$
\mathbf{h}=\tanh(\mathbf{W}\mathbf{x}+\mathbf{b})
$$

现在训练普通深层前馈网络时，隐藏层通常优先选择 ReLU；但理解 Tanh 很重要，因为它在循环神经网络、门控结构和一些归一化输出场景里仍然会出现。
