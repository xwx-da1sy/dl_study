# 07 特征图

前面已经学习了卷积核、多通道卷积和多卷积核计算。

这些内容里反复出现了一个词：特征图。

这一节专门把特征图讲清楚。

这一节仍然不写代码，只理解概念、来源、形状和作用。

## 1. 特征图是什么

特征图英文叫 feature map。

可以先这样理解：

```text
特征图 = 卷积核在整张图上寻找某种特征后得到的结果图
```

卷积核在图片上滑动。

每滑到一个位置，就算出一个数字。

这些数字按位置重新排成一张二维表，这张二维表就是特征图。

所以特征图不是原始图片，而是模型从原始图片里提取出来的一种信息。

## 2. 特征图从哪里来

特征图来自卷积计算。

一个卷积核在输入图像或输入特征图上滑动。

每个位置都会得到一个输出值。

很多输出值排成一张图，就得到一张特征图。

可以写成：

```text
输入图像 -> 卷积核滑动计算 -> 特征图
```

如果有多个卷积核，就会得到多张特征图。

这些特征图叠在一起，就形成多个输出通道。

## 3. 特征图里的数字表示什么

特征图里的每个数字，可以理解成某个位置对某种特征的响应程度。

如果某个位置的值比较大，说明这个卷积核关注的特征在这个位置比较明显。

如果某个位置的值比较小，说明这个特征在这个位置不明显。

比如一个卷积核比较关注竖线。

那么它输出的特征图里，竖线明显的位置可能会有较大的响应值。

所以可以这样理解：

```text
特征图记录的是：某种特征在不同位置出现得明显不明显。
```

## 4. 特征图和原图有什么区别

原图记录的是像素信息。

比如灰度图中，每个位置的数字表示明暗。

特征图记录的是特征响应。

比如某个位置像不像边缘，像不像拐角，像不像某种纹理。

所以：

```text
原图：更接近人眼看到的像素。
特征图：更接近模型提取出来的线索。
```

CNN 不是直接一直看原图。

它会一层一层把原始像素变成更适合分类的特征图。

## 5. 一个卷积核和一张特征图

前面学过：

```text
一个卷积核 -> 一张特征图
```

这是理解特征图最关键的一句话。

一个卷积核只负责产生一张特征图。

这张特征图表示：这个卷积核关注的特征在各个位置的响应情况。

如果有 16 个卷积核，就会得到 16 张特征图。

这 16 张特征图叠起来，就形成 16 个输出通道。

## 6. 特征图和通道的关系

在 CNN 里，经常把一张特征图看成一个通道。

比如卷积层输出形状是：

$$
16\times28\times28
$$

意思是：

- 有 16 张特征图。
- 每张特征图大小是 28 x 28。
- 也可以说输出有 16 个通道。

所以在卷积层输出中：

```text
输出通道数 = 特征图张数 = 卷积核个数
```

## 7. 特征图的高和宽由什么决定

特征图有两个方面需要看：

第一，特征图有多少张。

这由卷积核个数决定。

第二，每张特征图有多大。

这由输入大小、卷积核大小、padding 和 stride 决定。

也就是前面学过的公式：

$$
H_{out}=\left\lfloor\frac{H_{in}+2P-K}{S}\right\rfloor+1
$$

$$
W_{out}=\left\lfloor\frac{W_{in}+2P-K}{S}\right\rfloor+1
$$

所以：

```text
卷积核个数决定特征图张数。
卷积计算参数决定每张特征图的高和宽。
```

## 8. 浅层特征图和深层特征图

CNN 是一层一层提取特征的。

靠前的卷积层离原始图片更近。

这些浅层特征图通常更关注简单特征，比如边缘、线条、明暗变化。

靠后的卷积层经过了多次特征组合。

这些深层特征图通常更关注复杂特征，比如局部形状、笔画组合、物体部件。

可以先这样理解：

```text
浅层特征图：更像基础线索。
深层特征图：更像组合后的高级线索。
```

这也是 CNN 能处理图像的重要原因：它不是一步看懂整张图，而是逐层组合特征。

## 9. 用 MNIST 理解特征图

假设输入是一张 MNIST 手写数字图片：

$$
1\times28\times28
$$

第一层使用 16 个卷积核，并且让高宽保持不变。

那么输出可以是：

$$
16\times28\times28
$$

这不是说原图变成了 16 张新图片。

更准确地说，是模型从原图中提取出了 16 张特征图。

每张特征图都可以理解成一种特征的响应分布。

比如有的特征图可能对横线更敏感，有的可能对竖线更敏感，有的可能对拐角更敏感。

这些具体敏感什么，不是我们手工规定，而是模型训练出来的。

## 10. 特征图为什么还可以继续卷积

卷积层的输入不一定非得是原始图片。

它也可以接收上一层输出的特征图。

比如第一层输出：

$$
16\times28\times28
$$

那么第二层卷积看到的输入通道数就是 16。

第二层卷积核就要覆盖这 16 个输入通道。

如果第二层有 32 个卷积核，那么它会输出 32 张新的特征图。

也就是：

$$
16\times28\times28 \rightarrow 32\times H_{out}\times W_{out}
$$

这就是 CNN 中通道数经常逐层增加的原因。

## 11. 特征图和激活函数

卷积层算出来的结果通常还会经过激活函数，比如 ReLU。

前面学 MLP 时已经知道，ReLU 可以增加非线性。

在 CNN 中也是一样。

卷积层先得到特征图。

ReLU 再对特征图里的每个位置做处理。

可以理解成：

```text
卷积层负责提取特征。
激活函数负责让特征表达更灵活。
```

经过激活函数后，形状通常不变，仍然是一组特征图。

## 12. 特征图和池化层

池化层处理的通常也是特征图。

卷积层得到特征图后，池化层可以对特征图进行压缩。

比如一张 28 x 28 的特征图，经过某种池化后，可能变成 14 x 14。

这时特征图的张数通常不变，但每张特征图的高和宽变小。

可以这样理解：

```text
卷积层：产生特征图。
池化层：压缩特征图。
```

池化层的具体规则后面再单独学习。

## 13. 特征图不是越多越好

特征图越多，说明这一层输出通道越多。

输出通道多，模型能保留的特征种类通常更多。

但是特征图越多，也会带来更多计算量和参数量。

所以特征图数量不是越多越好。

它需要和任务复杂度、数据量、计算资源一起考虑。

入门阶段先记住：

```text
特征图数量增加，表达能力可能增强，但计算成本也会增加。
```

## 14. 容易混淆的地方

第一，特征图不是原图的复制。

它是卷积核从输入中提取出来的特征响应。

第二，输出通道不是颜色通道。

RGB 输入的 3 个通道是颜色信息，但卷积后的 16 个通道通常是 16 张特征图。

第三，特征图的数量和特征图的大小不是一回事。

特征图数量由卷积核个数决定。

每张特征图的高宽由卷积核大小、padding、stride 等决定。

第四，后一层看到的不是原图，而是前一层提取出来的特征图。

## 15. 本节小结

这一节先记住这些规则：

1. 特征图是卷积核在输入上滑动计算后得到的结果图。
2. 特征图里的数表示某种特征在不同位置的响应程度。
3. 一个卷积核产生一张特征图。
4. 多个卷积核产生多张特征图。
5. 特征图张数等于输出通道数。
6. 特征图的高宽由卷积核大小、padding、stride 等决定。
7. 浅层特征图偏向简单特征，深层特征图偏向组合特征。
8. 后一层卷积可以继续处理前一层输出的特征图。
