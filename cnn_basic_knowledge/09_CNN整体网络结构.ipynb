# 09 CNN 整体网络结构

前面已经分别学习了卷积层、padding、stride、多通道卷积、多卷积核、特征图和池化层。

这些内容都是 CNN 的零件。

这一节要把这些零件连起来，形成一个完整的 CNN 网络结构。

这一节仍然不写代码，只理解：**一张图片进入 CNN 后，通常会按什么顺序流动，形状会怎么变化。**

## 1. 为什么要学整体结构

前面单独学卷积层时，我们知道卷积层负责提取局部特征。

单独学池化层时，我们知道池化层负责压缩特征图。

但是在真实 CNN 中，这些层不是孤立存在的。

它们会按照一定顺序组合起来。

所以现在要从“单个零件”进入“完整网络”。

可以先记住一句话：

```text
CNN 的前半部分负责提取图像特征，后半部分负责根据特征完成分类。
```

## 2. 一个经典 CNN 的大致流程

一个入门级 CNN 通常可以写成：

```text
输入图片
-> 卷积层
-> 激活函数
-> 池化层
-> 卷积层
-> 激活函数
-> 池化层
-> 展平
-> 全连接层
-> 输出类别
```

这不是唯一结构，但很适合入门。

其中：

- 卷积层负责提取特征。
- 激活函数负责增加非线性。
- 池化层负责压缩特征图。
- 展平负责把多维特征图变成一维向量。
- 全连接层负责做最终分类。

## 3. 卷积层在整体结构中的作用

卷积层是 CNN 的特征提取核心。

它通过多个卷积核，从输入图片或上一层特征图中提取局部特征。

比如第一层卷积可能提取简单边缘、线条、明暗变化。

后面的卷积层可能继续组合这些简单特征，形成更复杂的局部形状。

可以这样理解：

```text
卷积层不是直接给出分类结果，而是先帮模型看出图像里有什么线索。
```

## 4. 激活函数在 CNN 中的作用

卷积层本质上还是线性计算。

如果只堆卷积层而没有激活函数，模型表达能力会受限制。

所以卷积层后面通常会接激活函数，比如 ReLU。

前面学习 MLP 时已经知道，ReLU 的作用是增加非线性。

在 CNN 中也是一样。

它不会改变特征图的形状，但会改变特征图里的数值。

可以先记住：

```text
卷积层提取特征，ReLU 让特征表达更灵活。
```

## 5. 池化层在整体结构中的作用

池化层通常接在卷积层和激活函数之后。

它会压缩每张特征图的高和宽。

比如：

```text
28 x 28 -> 14 x 14
14 x 14 -> 7 x 7
```

池化层一般不会改变通道数。

所以如果输入是 16 x 28 x 28，经过 2 x 2、stride = 2 的池化后，通常变成：

$$
16\times14\times14
$$

可以这样理解：

```text
池化层让特征图变小，让后面的计算更轻。
```

## 6. 为什么需要展平 Flatten

经过卷积层和池化层后，数据通常还是三维结构：

$$
C\times H\times W
$$

比如：

$$
32\times7\times7
$$

但是全连接层接收的通常是一维向量。

所以在进入全连接层之前，需要把特征图展平成一维向量。

例如：

$$
32\times7\times7=1568
$$

也就是：

```text
32 x 7 x 7 -> 1568
```

这一步叫 Flatten，中文可以叫展平。

## 7. 全连接层在 CNN 中的作用

CNN 前面的卷积层和池化层主要负责提取特征。

但最终还要做分类。

所以展平之后，通常会接全连接层。

全连接层拿到的是一维特征向量。

它会根据这些特征，输出每个类别的分数。

以 MNIST 为例，类别是 0 到 9，一共有 10 类。

所以最终输出通常是 10 个分数。

可以这样理解：

```text
卷积和池化负责看图像特征。
全连接层负责根据特征做判断。
```

## 8. 用 MNIST 看一遍形状变化

假设输入是一张 MNIST 图片：

$$
1\times28\times28
$$

第一层卷积使用 16 个卷积核，高宽保持不变：

$$
1\times28\times28 \rightarrow 16\times28\times28
$$

池化后高宽减半：

$$
16\times28\times28 \rightarrow 16\times14\times14
$$

第二层卷积使用 32 个卷积核，高宽保持不变：

$$
16\times14\times14 \rightarrow 32\times14\times14
$$

再池化一次：

$$
32\times14\times14 \rightarrow 32\times7\times7
$$

最后展平：

$$
32\times7\times7=1568
$$

再进入全连接层输出 10 个类别分数。

## 9. 加上 batch 后怎么表示

前面看的是单张图片。

训练时通常是一批图片一起输入。

如果 batch size 是 64，MNIST 输入形状是：

$$
64\times1\times28\times28
$$

经过第一层卷积后可能变成：

$$
64\times16\times28\times28
$$

经过池化后变成：

$$
64\times16\times14\times14
$$

batch 这个维度通常不变。

变化的是通道数、高度和宽度。

所以看 CNN 形状时，经常要盯住四个维度：

```text
batch x channel x height x width
```

## 10. CNN 前半部分和后半部分

入门阶段可以把 CNN 分成两部分。

第一部分是特征提取部分：

```text
卷积层 -> 激活函数 -> 池化层
```

这一部分的目标是把原始图片变成更有用的特征图。

第二部分是分类部分：

```text
展平 -> 全连接层 -> 输出类别
```

这一部分的目标是根据前面提取到的特征做最终判断。

所以 CNN 不是一层直接完成所有事情，而是先提特征，再分类。

## 11. CNN 和前面 MLP 的关系

CNN 的最后仍然可能使用全连接层。

所以 CNN 并不是完全抛弃 MLP。

区别在于，MLP 通常一开始就把图片展平。

CNN 则先保留图片的空间结构，通过卷积和池化提取特征，最后才展平。

可以这样对比：

```text
MLP：图片 -> 展平 -> 全连接分类

CNN：图片 -> 卷积/池化提特征 -> 展平 -> 全连接分类
```

所以 CNN 更适合图像任务，是因为它在分类前多了一段专门为图像设计的特征提取过程。

## 12. 容易混淆的地方

第一，卷积层不是最终分类层。

卷积层主要负责提取特征。

第二，池化层通常不会改变通道数。

它主要压缩每张特征图的高和宽。

第三，Flatten 不是学习层。

它只是改变形状，把多维特征图拉成一维向量。

第四，全连接层看到的不是原始图片，而是卷积和池化提取后的特征。

第五，输出类别数由任务决定。

MNIST 是 10 分类，所以最终输出 10 个分数。

## 13. 本节小结

这一节先记住这些规则：

1. CNN 通常由卷积层、激活函数、池化层、展平和全连接层组成。
2. 卷积层负责提取图像特征。
3. 激活函数负责增加非线性。
4. 池化层负责压缩特征图。
5. Flatten 负责把多维特征图展平成一维向量。
6. 全连接层负责根据特征做分类。
7. CNN 前半部分偏向特征提取，后半部分偏向分类判断。
8. CNN 相比 MLP 的关键优势，是在展平前先保留并利用图像空间结构。
