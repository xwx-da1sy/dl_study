# 14 CNN 图像分类案例总览

前面已经学完 CNN 的核心结构和训练闭环。

现在开始从“理论结构”过渡到“图像分类案例”。

这一节先不写代码，先把一个 CNN 图像分类案例拆开看清楚：

```text
数据从哪里来？
图片和标签怎么配对？
CNN 模型负责什么？
loss 和优化器放在哪里？
训练和测试分别在做什么？
```

这就像正式写代码前，先把地图摊开。

## 1. 为什么先学案例总览

如果直接进入代码，容易出现一个问题：

```text
每一行代码好像都认识，但不知道它们合起来在干什么。
```

所以在写 CNN 图像分类代码之前，先建立整体视角。

一个完整案例一般不是只有模型结构，它至少包括：

```text
数据
模型
损失函数
优化器
训练过程
测试过程
结果观察
```

后面真正写代码时，基本就是把这几个部分分别实现出来。

## 2. 图像分类案例的目标是什么

图像分类案例要解决的问题是：

```text
给模型一张图片，让模型判断它属于哪个类别。
```

比如 MNIST 手写数字识别：

```text
输入：一张 28 x 28 的灰度手写数字图片
输出：0 到 9 中的一个类别
```

所以这个案例里，模型不是生成图片，也不是找出图片中物体的位置。

它只做一件事：

```text
判断整张图片属于哪个类别。
```

## 3. 一个样本由什么组成

在监督学习里，一个训练样本通常由两部分组成：

```text
图片 + 标签
```

比如：

```text
图片：某张手写数字图片
标签：7
```

图片是模型的输入。

标签是标准答案。

模型先根据图片做预测，然后用预测结果和标签进行比较。

## 4. 数据集在案例中负责什么

数据集可以理解成：

```text
很多个 图片 + 标签 的集合。
```

比如 MNIST 训练集里有很多手写数字图片，每张图片都有对应标签。

数据集负责提供原始学习材料。

但是训练时通常不会一次把所有图片都塞给模型，而是分成很多小批次。

这就是 batch。

## 5. DataLoader 在案例中负责什么

DataLoader 可以先理解成“发牌的人”。

数据集里有很多图片，DataLoader 每次取出一批：

```text
一批图片
一批标签
```

然后交给模型训练。

所以可以这样记：

```text
Dataset：保存数据和标签。
DataLoader：按 batch 把数据送出来。
```

这一节只需要知道它们在流程里的位置，不需要急着背参数。

## 6. 模型在案例中负责什么

CNN 模型负责把图片变成类别分数。

对于 10 分类任务，模型最后会输出 10 个分数。

可以理解成：

```text
输入：一批图片
输出：每张图片属于各个类别的分数
```

比如一张图片经过模型后，模型可能认为：

```text
像 0 的程度
像 1 的程度
像 2 的程度
...
像 9 的程度
```

分数最高的类别，就是模型当前最倾向的预测。

## 7. CNN 模型内部大概分成两段

CNN 图像分类模型通常可以分成两段：

```text
特征提取部分
分类判断部分
```

特征提取部分一般由卷积层、激活函数、池化层组成。

它负责从图片中提取边缘、纹理、形状等特征。

分类判断部分一般由 Flatten 和全连接层组成。

它负责根据提取到的特征，输出类别分数。

## 8. loss 在案例中负责什么

模型输出类别分数后，需要和真实标签比较。

loss 负责衡量：

```text
模型这一次预测错得多不多。
```

如果模型把数字 7 预测成 7，而且很有把握，loss 通常会比较小。

如果模型把数字 7 预测成 1，而且很有把握，loss 通常会比较大。

所以 loss 是训练的信号来源。

没有 loss，模型就不知道自己应该往哪里改。

## 9. 优化器在案例中负责什么

反向传播会算出梯度。

优化器会根据梯度修改参数。

在 CNN 中，主要会修改：

```text
卷积层的权重和 bias
全连接层的权重和 bias
```

池化层、Flatten、普通 ReLU 通常没有可学习参数，所以它们不负责被优化器修改。

但反向传播时，梯度仍然会经过它们继续往前传。

## 10. 训练过程在案例中怎么走

训练过程可以先记成下面这条线：

```text
取出一个 batch 的图片和标签
-> 图片送入 CNN
-> 得到类别分数
-> 类别分数和标签计算 loss
-> 反向传播计算梯度
-> 优化器更新参数
-> 进入下一个 batch
```

这个过程会重复很多次。

每重复一次，模型的参数就有机会变得更合适一点。

## 11. 测试过程在案例中怎么走

测试过程和训练过程很像，但少了两件事：

```text
不反向传播
不更新参数
```

测试时只做：

```text
图片送入模型
-> 得到预测结果
-> 和真实标签比较
-> 统计准确率
```

也就是说，测试集更像考试。

考试时只看模型会不会，不再让它当场改答案。

## 12. 一个 CNN 图像分类案例的完整组成

现在可以把案例结构整理成：

```text
1. 准备数据集
2. 用 DataLoader 按 batch 读取数据
3. 定义 CNN 模型
4. 定义损失函数
5. 定义优化器
6. 训练模型
7. 测试模型
8. 观察 loss 和准确率
```

这就是后面 CNN-MNIST 案例的大框架。

## 13. 和前面 MLP-MNIST 案例的关系

你前面已经学过 MLP 做 MNIST 分类。

CNN-MNIST 和 MLP-MNIST 的任务目标一样：

```text
输入手写数字图片，输出数字类别。
```

训练流程也一样：

```text
前向传播
-> 计算 loss
-> 反向传播
-> 优化器更新参数
```

不同点主要在模型结构：

```text
MLP：先展平图片，再用全连接层分类。
CNN：先用卷积层提取图像特征，再分类。
```

所以 CNN-MNIST 不是从零开始的新任务，而是在原来的 MNIST 任务上换了一种更适合图像的模型。

## 14. 本节小结

这一节先记住：

```text
一个 CNN 图像分类案例 = 数据 + 模型 + loss + 优化器 + 训练 + 测试。
```

其中：

```text
数据提供图片和标签。
模型负责输出类别分数。
loss 衡量预测错误程度。
反向传播计算梯度。
优化器修改可学习参数。
测试集检查模型效果。
```

后面写代码时，不要把代码看成一堆 API。

要把每一部分放回这张流程图里理解。
