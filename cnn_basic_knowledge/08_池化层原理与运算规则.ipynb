# 08 池化层原理与运算规则

前面已经学习了卷积层会产生特征图。

卷积层负责提取特征，但如果一直只做卷积，特征图可能会越来越多，计算量也会越来越大。

所以 CNN 里经常会在卷积层后面接一个池化层。

这一节学习池化层：**它是什么，为什么要用，它是怎么压缩特征图的。**

## 1. 池化层是什么

池化层英文叫 pooling layer。

它的作用可以先理解成：

```text
对特征图做压缩，保留重要信息，减少计算量。
```

卷积层会得到特征图。

池化层通常接在卷积层后面，对这些特征图进行下采样。

下采样可以先理解成：把大的特征图变小。

比如：

```text
28 x 28 -> 14 x 14
32 x 32 -> 16 x 16
```

这就是池化层最直观的效果。

## 2. 为什么需要池化层

池化层主要有三个作用。

第一，减少计算量。

特征图变小后，后面的卷积层或全连接层要处理的数据就少了。

第二，保留主要特征。

池化不是随便丢信息，而是希望保留一个局部区域里比较有代表性的值。

第三，让模型对小范围位置变化不那么敏感。

比如手写数字稍微往左偏一点、往右偏一点，人还是能看出来它是同一个数字。

池化层有一点类似这个思想：不要过分依赖特征出现的精确位置，而是关注某个小区域里有没有明显特征。

## 3. 池化层处理的是什么

池化层通常处理的是特征图，而不是原始图片。

卷积层先从图片中提取特征，得到一组特征图。

池化层再对每张特征图分别进行压缩。

可以写成：

```text
输入图片 -> 卷积层 -> 特征图 -> 池化层 -> 更小的特征图
```

注意，池化层一般不会改变特征图的张数。

它主要改变每张特征图的高和宽。

## 4. 池化窗口是什么

池化层也有一个小窗口。

这个窗口会在特征图上滑动。

比如最常见的是 2 x 2 池化窗口。

它每次看特征图里的一个 2 x 2 小区域，然后把这个小区域压缩成一个数字。

可以这样理解：

```text
2 x 2 的局部区域 -> 1 个输出值
```

所以池化层会让特征图变小。

## 5. 最大池化 Max Pooling

最大池化是最常见的池化方式之一。

它的规则很简单：

```text
在池化窗口覆盖的区域里，取最大值作为输出。
```

比如一个 2 x 2 区域是：

$$
\begin{bmatrix}
1 & 3 \\
2 & 4
\end{bmatrix}
$$

最大值是 4。

所以最大池化输出就是：

$$
4
$$

最大池化关注的是：这个局部区域里最明显的特征是什么。

## 6. 平均池化 Average Pooling

平均池化也是一种池化方式。

它的规则是：

```text
在池化窗口覆盖的区域里，取平均值作为输出。
```

还是看这个 2 x 2 区域：

$$
\begin{bmatrix}
1 & 3 \\
2 & 4
\end{bmatrix}
$$

平均值是：

$$
\frac{1+3+2+4}{4}=2.5
$$

所以平均池化输出就是 2.5。

平均池化关注的是：这个局部区域整体大概是什么水平。

## 7. 最大池化和平均池化的区别

最大池化和平均池化都能压缩特征图。

区别在于它们保留信息的方式不同。

最大池化保留局部区域里最强的响应。

平均池化保留局部区域的平均水平。

可以这样记：

```text
最大池化：看最明显的特征。
平均池化：看整体平均情况。
```

在很多入门 CNN 结构中，最大池化更常见。

## 8. 池化层的 stride

池化层也有 stride。

它表示池化窗口每次移动几格。

最常见的组合是：

```text
池化窗口大小：2 x 2
stride：2
```

这表示每次看一个 2 x 2 区域，然后移动 2 个像素。

这样特征图的高和宽通常会减半。

比如：

```text
28 x 28 -> 14 x 14
14 x 14 -> 7 x 7
```

所以 2 x 2、stride = 2 是非常常见的池化设置。

## 9. 池化尺寸怎么计算

池化后的高宽计算方式和卷积很像。

如果输入高是 $H_{in}$，池化窗口大小是 $K$，padding 是 $P$，stride 是 $S$，那么输出高度是：

$$
H_{out}=\left\lfloor\frac{H_{in}+2P-K}{S}\right\rfloor+1
$$

宽度同理：

$$
W_{out}=\left\lfloor\frac{W_{in}+2P-K}{S}\right\rfloor+1
$$

入门阶段常见池化不加 padding，也就是 $P=0$。

最常见的是 2 x 2 池化，stride = 2。

## 10. 池化尺寸例子

假设输入特征图大小是：

$$
28\times28
$$

池化窗口大小是 2 x 2，所以：

$$
K=2
$$

不加 padding：

$$
P=0
$$

stride = 2：

$$
S=2
$$

代入公式：

$$
H_{out}=\frac{28+2\times0-2}{2}+1=14
$$

宽度同样是 14。

所以输出大小是：

$$
14\times14
$$

## 11. 池化会不会改变通道数

一般情况下，池化层不会改变通道数。

它会对每一张特征图分别做池化。

比如输入是：

$$
16\times28\times28
$$

表示有 16 张特征图，每张大小是 28 x 28。

经过 2 x 2、stride = 2 的池化后，输出通常变成：

$$
16\times14\times14
$$

可以看到，通道数还是 16。

变化的是每张特征图的高和宽。

## 12. 池化层有没有参数

池化层通常没有可学习参数。

卷积层有卷积核权重，这些权重会通过训练不断更新。

但池化层一般只是按照固定规则取最大值或平均值。

所以它不需要学习权重。

可以这样对比：

```text
卷积层：有参数，负责学习特征。
池化层：通常无参数，负责压缩特征图。
```

这也是池化层比较好理解的地方：它不是学习新权重，而是对已有特征图做整理和压缩。

## 13. 池化和卷积有什么区别

卷积层和池化层都会用小窗口在特征图上滑动。

但它们做的事情不同。

卷积层的小窗口里有可学习参数。

它的目标是提取特征。

池化层的小窗口里没有可学习参数。

它的目标是压缩特征图。

可以这样记：

```text
卷积：学习特征。
池化：压缩特征。
```

所以池化层通常不会单独用来学习复杂模式，而是配合卷积层使用。

## 14. 用 MNIST 理解池化

假设 MNIST 图片经过第一层卷积后，得到 16 张特征图：

$$
16\times28\times28
$$

如果接一个 2 x 2、stride = 2 的最大池化层，输出就是：

$$
16\times14\times14
$$

这表示：

- 特征图数量没有变，还是 16 张。
- 每张特征图变小了，从 28 x 28 变成 14 x 14。
- 后面的计算量减少了。

这就是 CNN 中常见的节奏：先卷积提特征，再池化压缩特征图。

## 15. 池化的一个直觉

最大池化可以理解成：只保留一个局部区域里最明显的信号。

比如某个 2 x 2 区域里，只要有一个位置强烈响应，就说明这个区域可能出现了某个重要特征。

这时最大池化会把这个强响应保留下来。

所以最大池化不太关心这个特征在 2 x 2 区域里的精确位置。

它更关心：

```text
这个小区域里有没有明显特征。
```

这可以让模型对小范围的位置移动更稳一点。

## 16. 本节小结

这一节先记住这些规则：

1. 池化层用于压缩特征图，减少计算量。
2. 池化层通常接在卷积层后面。
3. 最大池化取局部区域最大值。
4. 平均池化取局部区域平均值。
5. 常见池化窗口是 2 x 2，stride = 2。
6. 池化通常会减小特征图的高和宽。
7. 池化一般不会改变通道数。
8. 池化层通常没有可学习参数。
