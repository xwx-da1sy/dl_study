# 01 图像基础知识

从这一节开始学习 CNN，也就是卷积神经网络。

但是先不要急着看卷积，也不要急着写 PyTorch API。

CNN 主要用来处理图像，所以第一步要先弄清楚：**图像在计算机里到底是什么。**

## 1. 为什么先学图像基础

前面学习 MLP 时，我们已经知道：

$$
模型接收数字，经过计算，输出结果。
$$

那图片怎么输入给模型呢？

人看到图片，会看到数字、猫、狗、汽车、人脸。

但是计算机看到图片，本质上看到的是一堆数字。

所以学习 CNN 之前，要先把这件事想清楚：

$$
图片不是神秘的东西，图片就是按位置排列好的数字。
$$

CNN 后面要做的事情，就是从这些数字里学习图像特征。

## 2. 什么是像素

一张图片可以看成由很多小格子组成。

每一个小格子，就是一个像素。

如果一张图片大小是：

$$28\times 28$$

意思就是：

- 高度方向有 28 个像素。
- 宽度方向有 28 个像素。
- 一共有 $28\times28=784$ 个像素。

所以前面 MNIST 手写数字图片的大小是 28x28，本质上就是一张由 784 个小格子组成的图片。

## 3. 灰度图是什么

MNIST 是灰度图。

灰度图可以先理解成黑白照片，但它不是只有纯黑和纯白，中间还有很多深浅不同的灰色。

在常见表示方式中，一个像素的取值范围是：

$$0\sim255$$

含义是：

- 0 表示黑色。
- 255 表示白色。
- 中间的数表示不同程度的灰色。

例如：

```text
0     很黑
80    深灰
160   浅灰
255   很白
```

所以一张 28x28 的灰度图，可以看成一个 28 行、28 列的数字表。

## 4. 用矩阵理解灰度图

一张很小的灰度图可以这样理解：

```text
[[  0,   0,  30,  80],
 [  0,  50, 180, 220],
 [ 20, 160, 255, 200],
 [  0,  40, 120,  60]]
```

这个矩阵里的每一个数字，表示对应位置像素的明暗程度。

数字越小，越接近黑色。

数字越大，越接近白色。

这就是为什么我们可以把图片输入神经网络：因为图片最终可以变成数字。

## 5. RGB 彩色图是什么

灰度图每个像素只有一个数字。

彩色图通常需要三个数字。

这三个数字分别表示：

- R：Red，红色分量。
- G：Green，绿色分量。
- B：Blue，蓝色分量。

一个彩色像素可以写成：

```text
(R, G, B)
```

例如：

```text
(255, 0, 0)     偏红
(0, 255, 0)     偏绿
(0, 0, 255)     偏蓝
(255, 255, 255) 白色
(0, 0, 0)       黑色
```

所以彩色图比灰度图多了一个东西：颜色通道。

## 6. 什么是通道

通道可以先理解成一张图片的不同信息层。

灰度图只有一层明暗信息，所以是 1 个通道。

RGB 彩色图有三层颜色信息，所以是 3 个通道。

可以这样记：

```text
灰度图：1 个通道
RGB 图：3 个通道
```

如果一张彩色图片大小是 32x32，那么它不是只有一个 32x32 的数字表，而是有 3 个 32x32 的数字表：

```text
R 通道：32 x 32
G 通道：32 x 32
B 通道：32 x 32
```

合在一起，就是一张彩色图片。

## 7. 图像的形状怎么表示

在深度学习中，图像通常要关心三个维度：

```text
通道数 x 高度 x 宽度
```

也就是：

$$C\times H\times W$$

其中：

- $C$ 表示 channel，通道数。
- $H$ 表示 height，高度。
- $W$ 表示 width，宽度。

MNIST 是 28x28 的灰度图，所以可以写成：

$$1\times28\times28$$

CIFAR-10 是 32x32 的 RGB 彩色图，所以可以写成：

$$3\times32\times32$$

## 8. 加上 batch 后是什么形状

前面讲的是一张图片。

训练神经网络时，通常不是一张一张输入，而是一批一批输入。

如果一次输入 $B$ 张图片，那么形状前面会多一个 batch 维度：

$$B\times C\times H\times W$$

例如一次输入 64 张 MNIST 图片：

$$64\times1\times28\times28$$

一次输入 64 张 CIFAR-10 图片：

$$64\times3\times32\times32$$

以后学习 CNN 时，一定要经常盯住这个形状。图像模型很多问题，本质上都是 shape 没想清楚。

## 9. MLP 怎么处理图片

前面做 MNIST 时，MLP 不能直接处理 28x28 的二维图片。

所以我们把图片展平成一维向量：

$$28\times28=784$$

也就是：

```text
1 x 28 x 28  ->  784
```

这样做的好处是简单。

但是它也有一个问题：图片原来的空间结构被打散了。

比如一个数字的笔画，原来是相邻像素连在一起形成的。展平之后，模型看到的是一长串数字，图片的上下左右关系没有那么直接了。

## 10. CNN 为什么要出现

CNN 的出发点很自然：

```text
既然图片有高、宽、通道这些结构，那模型最好也利用这些结构。
```

也就是说，CNN 不希望一开始就把图片完全展平。

它会先保留图像的空间结构，从局部区域里提取特征。

比如：

- 某一小块区域像不像一条边。
- 某一小块区域像不像一个角。
- 多个局部特征组合起来，像不像一个数字。

这就是后面要学习卷积层的原因。

## 11. 本节小结

这一节先记住几句话：

1. 图片在计算机里本质上是数字。
2. 一个像素就是图片里的一个小格子。
3. 灰度图每个像素通常用一个数字表示明暗。
4. RGB 彩色图每个像素通常用三个数字表示颜色。
5. 图像在深度学习中常写成 $C\times H\times W$。
6. 加上 batch 后，常写成 $B\times C\times H\times W$。
7. MLP 会把图片展平，CNN 会先保留图片结构。
