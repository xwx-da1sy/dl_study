# 11 LeNet 经典网络结构

前面已经学习了 CNN 的核心零件：卷积层、池化层、特征图、Flatten 和全连接分类头。

这一节开始看第一个经典 CNN 网络：LeNet。

LeNet 很适合入门，因为它结构简单，而且本来就和手写数字识别任务关系很近。

这一节仍然不写代码，只理解 LeNet 的结构和形状变化。

## 1. LeNet 是什么

LeNet 是早期非常经典的卷积神经网络结构。

它最常被用来讲解手写数字识别。

从学习角度看，LeNet 的价值不在于它现在有多强，而在于它把 CNN 的基本结构讲得很清楚。

可以先这样理解：

```text
LeNet = 卷积层 + 池化层 + 卷积层 + 池化层 + 全连接分类头
```

它正好把前面学过的 CNN 零件串成了一个完整网络。

## 2. 为什么先学 LeNet

后面还会遇到更复杂的 CNN，比如 AlexNet、VGG、ResNet。

但一开始直接看那些网络，很容易被层数和结构绕晕。

LeNet 的结构比较短，适合用来理解 CNN 的主线：

```text
先提取简单特征
再压缩特征图
再提取更复杂特征
再压缩特征图
最后展平并分类
```

所以 LeNet 是一个很好的过渡：从 CNN 概念，走向 CNN 实战。

## 3. LeNet 的整体结构

入门阶段可以把 LeNet 理解成下面这个顺序：

```text
输入图片
-> 卷积层 C1
-> 激活函数
-> 池化层 S2
-> 卷积层 C3
-> 激活函数
-> 池化层 S4
-> Flatten 展平
-> 全连接层
-> 输出层
```

其中 C 通常表示 convolution，也就是卷积层。

S 可以理解成 subsampling，也就是下采样，和池化的思想接近。

现在先不纠结历史版本中的细节，只抓住结构主线。

## 4. LeNet 的第一段：卷积提特征

第一层卷积负责从原始图片中提取低级特征。

如果输入是手写数字图片，这一层可能会学习到一些简单线索，比如边缘、笔画方向、明暗变化。

假设输入是 MNIST 图片：

$$
1\times28\times28
$$

如果第一层使用 6 个卷积核，输出就会有 6 张特征图。

也就是：

$$
6\times H_{out}\times W_{out}
$$

这里的 6 表示模型从原图中提取了 6 类基础特征。

## 5. LeNet 的第二段：池化压缩特征图

卷积层得到特征图后，池化层会对特征图做压缩。

池化层通常不会改变特征图张数。

它主要减小每张特征图的高和宽。

比如：

```text
6 张较大的特征图 -> 6 张较小的特征图
```

这个过程可以减少计算量，也让模型不要太依赖特征出现的精确位置。

所以第一段卷积和池化可以理解成：

```text
先从原图里找基础特征，再把特征图压缩一下。
```

## 6. LeNet 的第三段：再次卷积

第一段卷积看到的是原始图片。

第二个卷积层看到的不是原图，而是前面输出的特征图。

这意味着第二个卷积层是在已有特征的基础上继续提取更复杂的特征。

比如第一层可能提取了边缘和简单笔画。

第二层可能会把这些简单特征组合成更复杂的局部形状。

可以这样理解：

```text
第一层卷积：看基础线索。
第二层卷积：组合基础线索，形成更复杂的线索。
```

## 7. LeNet 的第四段：再次池化

第二个卷积层之后，通常还会再接一个池化层。

这样可以继续压缩特征图的高和宽。

经过两次卷积和两次池化后，特征图已经比原图小很多。

但这些特征图通常包含了更有用的图像线索。

可以这样理解：

```text
越往后，空间尺寸通常变小，特征表达通常更抽象。
```

这就是 CNN 逐层提取特征的味道。

## 8. LeNet 的后半段：展平和全连接分类

卷积和池化结束后，数据仍然是特征图形式。

比如可能是：

$$
16\times5\times5
$$

这时需要先 Flatten 展平：

$$
16\times5\times5=400
$$

然后把这个一维特征向量送入全连接层。

全连接层根据这些特征进行最终分类。

如果任务是 MNIST，最后输出 10 个类别分数。

## 9. 用形状串起一个 LeNet 思路

为了和前面学过的 MNIST 衔接，可以先看一个简化的形状流动。

输入：

$$
1\times28\times28
$$

第一层卷积，得到 6 张特征图：

$$
6\times24\times24
$$

第一次池化，高宽减半：

$$
6\times12\times12
$$

第二层卷积，得到 16 张特征图：

$$
16\times8\times8
$$

第二次池化，高宽减半：

$$
16\times4\times4
$$

展平：

$$
16\times4\times4=256
$$

最后进入全连接层，输出 10 个类别分数。

这里的具体尺寸会因为输入大小、卷积核大小、padding 设置不同而变化。

入门阶段重点不是死记数字，而是理解形状变化方向。

## 10. LeNet 的核心节奏

LeNet 的核心节奏可以概括成：

```text
卷积 -> 池化 -> 卷积 -> 池化 -> 展平 -> 全连接 -> 输出
```

这个节奏后来影响了很多 CNN 结构。

虽然现代网络更深、更复杂，但很多基本思想仍然类似：

- 前面用卷积提取特征。
- 中间逐渐压缩空间尺寸。
- 后面根据特征完成分类。

所以学 LeNet，不只是学一个历史网络，而是学 CNN 的基本组织方式。

## 11. LeNet 和 MLP 的区别

前面 MLP 做 MNIST 时，一开始就把图片展平成 784 维向量。

也就是：

$$
1\times28\times28 \rightarrow 784
$$

LeNet 不会一开始就展平。

它会先经过卷积和池化，保留并利用图像的空间结构。

等到特征提取完成后，才 Flatten 进入全连接层。

可以这样对比：

```text
MLP：先展平，再分类。
LeNet：先提取图像特征，再展平分类。
```

这就是 CNN 相比普通 MLP 更适合图像任务的重要原因。

## 12. LeNet 中每类层的职责

可以把 LeNet 中的层分工整理成这样：

```text
卷积层：提取局部图像特征。
激活函数：增加非线性表达能力。
池化层：压缩特征图，减少计算量。
Flatten：把特征图展平成一维向量。
全连接层：根据特征完成分类判断。
```

这几类层配合起来，就形成了一个完整的 CNN 分类模型。

## 13. 容易混淆的地方

第一，LeNet 不是只有卷积层。

它还包含池化层和全连接层。

第二，池化层通常不会增加特征图张数。

它主要压缩每张特征图的高和宽。

第三，全连接层不是直接处理原始图片。

它处理的是前面卷积和池化得到的特征。

第四，LeNet 的具体数字不一定要死背。

更重要的是理解结构顺序和形状变化。

## 14. 本节小结

这一节先记住这些规则：

1. LeNet 是经典 CNN 结构，适合用来入门理解图像分类。
2. LeNet 的基本节奏是卷积、池化、卷积、池化、展平、全连接。
3. 卷积层负责提取图像特征。
4. 池化层负责压缩特征图。
5. 后面的全连接层负责分类。
6. LeNet 不会一开始就展平图片，而是先提取图像特征。
7. 学 LeNet 的重点是理解 CNN 的组织方式，而不是死背某一组尺寸数字。
