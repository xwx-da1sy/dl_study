# 02 CNN 概述介绍

上一节我们先学习了图像基础知识：像素、灰度图、RGB 图、通道，以及图像的形状。

这一节开始正式认识 CNN。

先不急着写代码，也先不急着看 `Conv2d` 这些 API。

这一节只解决一个问题：**CNN 到底是什么，它为什么适合处理图像。**

## 1. CNN 是什么

CNN 的全称是 Convolutional Neural Network，中文叫卷积神经网络。

它本质上仍然是神经网络。

前面学过的 MLP 是神经网络，CNN 也是神经网络。

区别在于：CNN 里面加入了专门处理图像的结构，最典型的就是卷积层和池化层。

可以先这样理解：

```text
CNN = 更适合处理图像的神经网络
```

它不是完全推翻前面学过的内容，而是在前面 MLP、激活函数、损失函数、优化器、反向传播这些基础上，增加了适合图像的特征提取方式。

## 2. 为什么图像任务需要 CNN

前面做 MNIST 时，我们用 MLP 也能完成手写数字分类。

做法是先把图片展平成一维向量：

$$
1\times28\times28 \rightarrow 784
$$

这样确实可以训练模型。

但是它有一个明显问题：图片原来的空间结构被打散了。

比如一个手写数字 `7`，它的横线和斜线是由相邻像素组成的。

展平之后，这些上下左右的邻近关系不再那么直接。

也就是说，MLP 看到的是一长串数字，而 CNN 更像是在看一块一块的图像区域。

## 3. 图像有局部特征

图像和普通表格数据不太一样。

图像里，相邻像素之间通常有关系。

比如：

- 一小片连续变亮的区域，可能是一条边缘。
- 一条边缘和另一条边缘组合起来，可能形成一个角。
- 多个边、角、纹理组合起来，可能形成一个数字、眼睛、车轮或者猫耳朵。

所以理解 CNN 时，要先记住一个重要想法：

```text
图像特征通常是从局部区域一步一步组合出来的。
```

CNN 正是利用了图像的这个特点。

## 4. CNN 的核心思路

CNN 的核心思路可以先概括成三句话。

第一，先看局部。

CNN 不会一开始就把整张图片全部打散，而是用一个小窗口去观察图片的一小块区域。

第二，同一种特征可以在不同位置出现。

比如一条斜线，可能出现在图片左上角，也可能出现在右下角。CNN 希望同一个特征检测方式可以在整张图片上重复使用。

第三，逐层提取更复杂的特征。

浅层可能学习边缘、角点；更深层可能学习笔画、局部形状；再深一点可能组合成完整类别的特征。

所以 CNN 的学习过程可以粗略理解成：

```text
像素 -> 边缘 -> 局部形状 -> 整体类别
```

## 5. CNN 通常由哪些部分组成

一个入门级 CNN 通常可以看成由三类主要结构组成：

```text
卷积层 -> 池化层 -> 全连接层
```

当然中间通常还会配合激活函数，比如 ReLU。

先不用急着理解每一层的公式，先记作用。

卷积层：负责从图像中提取局部特征。

池化层：负责压缩特征图的大小，保留比较重要的信息。

全连接层：负责根据前面提取到的特征，输出最终分类结果。

也可以写成更完整一点：

```text
输入图片 -> 卷积提特征 -> 激活增加非线性 -> 池化压缩 -> 全连接分类 -> 输出类别
```

## 6. 卷积层先理解成什么

卷积层是 CNN 里最核心的部分。

但是这一节先不讲卷积计算细节。

现在先把卷积层理解成：

```text
在图片上滑动的小窗口，用来寻找某种局部特征。
```

比如这个小窗口可能在寻找：

- 有没有横线。
- 有没有竖线。
- 有没有斜线。
- 有没有边缘。
- 有没有某种局部纹理。

不同的卷积核可以学习不同的特征。

这个地方先有直觉就够，下一节再专门学习卷积层。

## 7. 池化层先理解成什么

池化层可以先理解成对特征图做压缩。

比如原来有一张比较大的特征图，池化后变小一些。

它的直观作用是：

- 减少后面的计算量。
- 保留比较明显的特征。
- 让模型对小范围的位置变化不那么敏感。

例如手写数字稍微向左偏一点、向右偏一点，人还是能看出它是同一个数字。

池化层就有一点这种味道：不要太纠结特征出现的精确位置，而是更关心某个区域里有没有明显特征。

## 8. CNN 和 MLP 的关系

CNN 不是把 MLP 扔掉了。

很多 CNN 的最后，仍然会接全连接层来做分类。

区别在于：

MLP 通常一开始就把图片展平，然后直接分类。

CNN 会先保留图像结构，用卷积层和池化层提取图像特征，最后再交给全连接层分类。

可以这样对比：

```text
MLP：图片 -> 展平 -> 全连接分类

CNN：图片 -> 卷积/池化提取特征 -> 全连接分类
```

所以 CNN 更适合图像，并不是因为它完全不同，而是因为它在分类前先做了更适合图像的特征提取。

## 9. CNN 输入和输出是什么

以 MNIST 为例，输入是一批灰度图片。

单张图片可以理解成：

$$
1\times28\times28
$$

如果 batch size 是 64，输入就是：

$$
64\times1\times28\times28
$$

输出仍然是分类结果。

MNIST 要识别 0 到 9，所以最后仍然输出 10 个类别分数。

也就是说，CNN 的输入保留图像结构，但输出目标和前面 MLP 做 MNIST 时是一样的：判断图片属于哪一类。

## 10. 这一节先不要急着掌握什么

这一节是 CNN 概述，不是卷积计算课。

所以现在先不要急着纠结：

- 卷积核具体怎么算。
- padding 怎么计算。
- stride 怎么影响尺寸。
- PyTorch 的 `Conv2d` 参数怎么写。

这些内容后面会一节一节拆开。

现在只需要先建立整体地图：

```text
CNN 是专门适合图像任务的神经网络。
它先提取图像特征，再根据特征完成分类。
```

## 11. 本节小结

这一节先记住几句话：

1. CNN 是卷积神经网络，本质上仍然是神经网络。
2. CNN 更适合图像，因为它会利用图像的空间结构。
3. 图像里的特征通常从局部区域开始出现。
4. 卷积层负责提取局部特征。
5. 池化层负责压缩特征图，降低计算量。
6. 全连接层负责根据提取到的特征输出分类结果。
7. CNN 可以理解成：先提取图像特征，再完成分类。
